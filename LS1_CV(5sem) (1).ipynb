1. Write a program to read, display and write an image using OpenCV in Python.

```python
import cv2
from google.colab.patches import cv2_imshow

# Read the image
image = cv2.imread("/content/Image.1.jpeg")

# Check if image is loaded successfully
if image is None:
    print("Image not found!")
else:
    # Resize the image to 50% of its original size
    small_image = cv2.resize(image, None, fx=0.5, fy=0.5)

    # Display the resized image
    cv2_imshow(small_image)

    # Save the resized image
    cv2.imwrite("/content/small_image.jpg", small_image)

    print("Image displayed and saved successfully.")
```

2. Write a program to convert a Color Image into Grayscale and Binary Image .

```python
# Convert color image to grayscale
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Display grayscale image
print("Grayscale Image: ")

cv2_imshow(gray_image)

# Convert grayscale image to binary image
_, binary_image = cv2.threshold(gray_image, 127, 255, cv2.THRESH_BINARY)

# Display binary image
print("Binary Image:")
cv2_imshow(binary_image)

# Save the images
cv2.imwrite("/content/grayscale_image.jpg", gray_image)
cv2.imwrite("/content/binary_image.jpg", binary_image)

print("Images converted and saved successfully.")
```

Grayscale: 0–255 shades of gray

Binary: only 0 (black) and 255 (white)

3. Write a program to Compute and Display the Complement (Negative) of the Image

```python
# Compute the complement (negative) of the image
negative_image = 255 - image

# Display negative image
print("Negative Image:")
cv2_imshow(negative_image)

# Save the negative image
cv2.imwrite("/content/negative_image.jpg", negative_image)

print("Negative image created and saved successfully.")
```

4. Write a Program to Resize, Crop and Rotate an Image.

```python
import cv2
import matplotlib.pyplot as plt

# Read the image
image = cv2.imread("/content/Image.1.jpeg")

# Check if image is loaded successfully
if image is None:
    print("Error: Image not found!")
else:
    # Convert BGR to RGB for displaying with matplotlib
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

    # ---------------- RESIZE ----------------
    resized = cv2.resize(image, (300, 200))
    resized_rgb = cv2.cvtColor(resized, cv2.COLOR_BGR2RGB)

    # ---------------- CROP ----------------
    # Format: image[y1:y2, x1:x2]
    cropped = image[300:600, 250:580]
    cropped_rgb = cv2.cvtColor(cropped, cv2.COLOR_BGR2RGB)

    # ---------------- ROTATE ----------------
    rotated = cv2.rotate(image, cv2.ROTATE_90_CLOCKWISE)
    rotated_rgb = cv2.cvtColor(rotated, cv2.COLOR_BGR2RGB)

    # Display all images
    plt.figure(figsize=(12, 8))

    plt.subplot(2, 2, 1)
    plt.imshow(image_rgb)
    plt.title("Original Image")
    plt.axis("off")

    plt.subplot(2, 2, 2)
    plt.imshow(resized_rgb)
    plt.title("Resized Image")
    plt.axis("off")

    plt.subplot(2, 2, 3)
    plt.imshow(cropped_rgb)
    plt.title("Cropped Image")
    plt.axis("off")

    plt.subplot(2, 2, 4)
    plt.imshow(rotated_rgb)
    plt.title("Rotated Image")
    plt.axis("off")

    plt.show()
```

5. Split any image and then merge .

```python
import cv2
import matplotlib.pyplot as plt
import numpy as np

# Read image
image = cv2.imread("/content/Image.1.jpeg")

# Split image
B, G, R = cv2.split(image)

# Create black channels
zeros = np.zeros_like(B)

# Create colored channel images
blue_image = cv2.merge([B, zeros, zeros])
green_image = cv2.merge([zeros, G, zeros])
red_image = cv2.merge([zeros, zeros, R])

# Merge all channels back
merged_image = cv2.merge([B, G, R])

# Convert to RGB for Matplotlib
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
blue_rgb = cv2.cvtColor(blue_image, cv2.COLOR_BGR2RGB)
green_rgb = cv2.cvtColor(green_image, cv2.COLOR_BGR2RGB)
red_rgb = cv2.cvtColor(red_image, cv2.COLOR_BGR2RGB)
merged_rgb = cv2.cvtColor(merged_image, cv2.COLOR_BGR2RGB)

# Display
plt.figure(figsize=(12, 8))

plt.subplot(2, 3, 1)
plt.imshow(image_rgb)
plt.title("Original Image")
plt.axis("off")

plt.subplot(2, 3, 2)
plt.imshow(blue_rgb)
plt.title("Blue Channel")
plt.axis("off")

plt.subplot(2, 3, 3)
plt.imshow(green_rgb)
plt.title("Green Channel")
plt.axis("off")

plt.subplot(2, 3, 4)
plt.imshow(red_rgb)
plt.title("Red Channel")
plt.axis("off")

plt.subplot(2, 3, 5)
plt.imshow(merged_rgb)
plt.title("Merged Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```
